## Проверка при помощи python библиотеки 

```python
import numpy as np
from scipy.optimize import minimize
import pandas as pd


# Определение целевой функции
def objective(x):
    return np.sum(x**2)

# Ограничение
# g(x) = 1 - sum(x_i^4) <= 0 \Rightarrow sum(x_i^4) <= 1
def constraint(x):
    return 1 - np.sum(x**4)

def objective_max(x):
    return -np.sum(x**2)
# Количество переменных
n = 5  # Можно изменить

# Начальное приближение (равномерное распределение)
x0 = np.ones(n) / np.sqrt(n)

# Ограничение в формате scipy.optimize
constraints = {'type': 'ineq', 'fun': constraint}

# Запуск оптимизации для минимизации
result = minimize(objective, x0, constraints=constraints)

# Вывод результатов
optimal_x = result.x
optimal_value = result.fun

# Формируем таблицу с результатами
df = pd.DataFrame({'x_i': np.round(optimal_x, 4)})
print(df)
np.round(optimal_value, 4)
```

```python
# считаем максимум
result_max = minimize(objective_max, x0, constraints=constraints, method='SLSQP')

# Вывод результатов
optimal_x = result_max.x
optimal_value = result_max.fun

df = pd.DataFrame({'x_i': np.round(optimal_x, 4)})
print(df)
np.round(-optimal_value, 4)
```

**Вывод:** Экстремумы максимума и минимума с аналитическими совпадают.
