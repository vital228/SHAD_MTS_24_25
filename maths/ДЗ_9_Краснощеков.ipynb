## Проверка при помощи python библиотеки 

```python
from scipy.optimize import linprog

c = [-2, -3, 0, 1, 0, 0]  # Меняем знак, так как scipy.minimize

# Матрица ограничений 
A_eq = [
    [2, -1, 0, -2, 1, 0],
    [3,  2, 1, -3, 0, 0],
    [-1, 3, 0,  4, 0, 1]
]

# свободные члены
b_eq = [16, 18, 24]

# все x >= 0
x_bounds = [(0, None)] * 6  

# Запуск симплекс-метода
result = linprog(c, A_eq=A_eq, b_eq=b_eq, bounds=x_bounds, method="highs")

# Вывод результатов
if result.success:
    print("x = ", result.x)
    print("L* = ", -result.fun)  # Минус для восстановления макс
else:
    print("Решение не найдено:", result.message)
```
